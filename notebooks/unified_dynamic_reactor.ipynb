## Phase 2 · Priority 1 — Urmès et al. (2019) Langmuir–Hinshelwood law

This notebook reconstructs the one-site Langmuir–Hinshelwood expressions in [Urmès et al. (2019), *Catalysts* 9(2), 180](https://doi.org/10.3390/catal9020180).

**Scope.** The paper studies tail-end acetylene hydrogenation over 0.05 wt.% Pd/α-Al₂O₃. Its sequence covers C₂H₂ → C₂H₄ and the follow-on C₂H₄ → C₂H₆ reaction. These are the paper's assumptions; they do not establish the same rate law for a later PdAg/CO model.

Notation: θᵢ is the fraction of sites occupied by species i; θ₍*₎ is the free-site fraction; Pᵢ is a gas-phase partial pressure; an adsorbed species is marked with *. Use pressure units consistent with each equilibrium constant.

The authors assume one shared site type, Langmuir adsorption, quasi-equilibrium for steps 1–3 and 5–6, and one-way rate-determining steps 4 and 7. Vinyl and ethyl intermediates are retained in the rate expressions but omitted from the reduced site balance as low-coverage species.

### 1. Elementary sequence

Urmès et al. use the seven-step Horiuti–Polanyi sequence below (Table 3):

| Step | Elementary reaction | Treatment |
|---|---|---|
| 1 | \(H_2 + 2* \rightleftharpoons 2H*\) | Quasi-equilibrium; dissociative hydrogen adsorption |
| 2 | \(C_2H_2 + * \rightleftharpoons C_2H_2*\) | Quasi-equilibrium acetylene adsorption |
| 3 | \(C_2H_2* + H* \rightleftharpoons C_2H_3* + *\) | Quasi-equilibrium first H addition |
| 4 | \(C_2H_3* + H* \rightarrow C_2H_4* + *\) | One-way rate-determining step for ethylene formation |
| 5 | \(C_2H_4 + * \rightleftharpoons C_2H_4*\) | Quasi-equilibrium ethylene adsorption/desorption |
| 6 | \(C_2H_4* + H* \rightleftharpoons C_2H_5* + *\) | Quasi-equilibrium first H addition |
| 7 | \(C_2H_5* + H* \rightarrow C_2H_6 + 2*\) | One-way rate-determining step for ethane formation |

Hydrogen consumes two neighboring vacant sites in step 1. Ethane is assumed to desorb immediately. The paper reports no observed C₄ products in these experiments. Source: [Urmès et al. (2019), Table 3 and pp. 3–4](https://doi.org/10.3390/catal9020180).

### 2. Quasi-equilibrium coverages

Dissociative hydrogen adsorption gives

\[
K_1 = \frac{\theta_H^2}{P_{H_2}\theta_*^2}
\quad\Longrightarrow\quad
\theta_H = \sqrt{K_1P_{H_2}}\,\theta_*.
\]

The square root appears because one gas-phase H₂ molecule forms two adsorbed H atoms.

For acetylene adsorption and the first H addition,

\[
K_2=\frac{\theta_{C_2H_2}}{P_{C_2H_2}\theta_*},
\qquad
K_3=\frac{\theta_{C_2H_3}\theta_*}{\theta_{C_2H_2}\theta_H}.
\]

Substitution yields \(\theta_{C_2H_2}=K_2P_{C_2H_2}\theta_*\) and \(\theta_{C_2H_3}=K_2K_3P_{C_2H_2}\sqrt{K_1P_{H_2}}\,\theta_*\).

The paper's \(K_5\) is written in pressure units, as the inverse of the usual ethylene adsorption constant:

\[
K_5=\frac{P_{C_2H_4}\theta_*}{\theta_{C_2H_4}},
\qquad
K_6=\frac{\theta_{C_2H_5}\theta_*}{\theta_{C_2H_4}\theta_H}.
\]

Therefore \(\theta_{C_2H_4}=(P_{C_2H_4}/K_5)\theta_*\) and \(\theta_{C_2H_5}=(K_6/K_5)P_{C_2H_4}\sqrt{K_1P_{H_2}}\,\theta_*\). The vinyl and ethyl coverages form the rates even though they are dropped from the reduced site balance.

Source: [Urmès et al. (2019), pp. 3–5](https://doi.org/10.3390/catal9020180).

### 3. Competitive site balance and squared denominator

The full balance is
\[
1=\theta_*+\theta_H+\theta_{C_2H_2}+\theta_{C_2H_3}
+\theta_{C_2H_4}+\theta_{C_2H_5}.
\]
Dropping the low-coverage vinyl and ethyl terms, then substituting the coverages above, gives
\[
1=\theta_*\left(1+\sqrt{K_1P_{H_2}}+K_2P_{C_2H_2}
+\frac{P_{C_2H_4}}{K_5}\right).
\]

Define
\[
D=1+\sqrt{K_1P_{H_2}}+K_2P_{C_2H_2}
+\frac{P_{C_2H_4}}{K_5};
\qquad \theta_*=\frac{1}{D}.
\]

Each rate-determining step consumes an intermediate and H*. Each coverage is proportional to \(\theta_*\), so their product is proportional to \(\theta_*^2=1/D^2\):
\[
r_4=k_4\theta_{C_2H_3}\theta_H
=k_4K_1K_2K_3\frac{P_{C_2H_2}P_{H_2}}{D^2},
\]
\[
r_7=k_7\theta_{C_2H_5}\theta_H
=k_7\frac{K_1K_6}{K_5}\frac{P_{C_2H_4}P_{H_2}}{D^2}.
\]

The paper multiplies by the surface-site inventory \(N_S\) and fits Arrhenius rate constants. Its final fitted \(k_4^0\) and \(k_7^0\) absorb \(K_3\) and \(K_6\), respectively:
\[
-r_{C_2H_2}=N_Sk_4^0e^{-E_4/(RT)}
\frac{K_1K_2P_{C_2H_2}P_{H_2}}{D^2},
\qquad
r_{C_2H_6}=N_Sk_7^0e^{-E_7/(RT)}
\frac{(K_1/K_5)P_{C_2H_4}P_{H_2}}{D^2}.
\]

Source: [Urmès et al. (2019), pp. 4–5](https://doi.org/10.3390/catal9020180).

### 4. The published \(K_1\) conflict — unresolved

The parameter-estimation discussion on p. 5 estimates \(K_1=611\ \mathrm{Pa^{-1}}\) from \(\sigma_s=0.16\), \(E_d=69\ \mathrm{kJ\,mol^{-1}}\), and a desorption prefactor of \(10^{13}\ \mathrm{s^{-1}}\), then says the value is fixed. Table 5 on p. 6 instead lists \(K_1=13\ \mathrm{Pa^{-1}}\), also fixed.

These are two different values with the same units in the same paper. The derivation does not explain a conversion or a changed definition. The source therefore leaves an internal inconsistency; neither value can be selected as the source-backed value without further evidence.

Because the denominator contains \(\sqrt{K_1P_{H_2}}\), the two values change that contribution by \(\sqrt{611/13}\approx6.86\) at the same hydrogen pressure. If a later implementation needs a provisional value, keep the alternatives as explicitly labeled sensitivity cases rather than silently choosing one.

Source: [Urmès et al. (2019), pp. 5–6, Table 5](https://doi.org/10.3390/catal9020180).
